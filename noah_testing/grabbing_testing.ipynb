```python
import subprocess
import stretch_body.robot
import time
```

```python
try:
    # This runs the terminal command and waits for it to finish.
    # The 'check=True' part will raise an error if the script fails,
    # which is good for safety.
    subprocess.run("stretch_free_robot_process.py", shell=True, check=True)
    subprocess.run("stretch_robot_home.py", shell=True, check=True)
    
    print("\n--- Homing Complete! ---")
    time.sleep(1) # Give a moment for everything to settle

except subprocess.CalledProcessError as e:
    print(f"Homing script failed with error: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```
